```python
# 1. List Creation
age_list = [24, 25, 26, 27, 28]
name_list = ["Anjali", "Meera", "Asha", "Divya", "Lakshmi"]

print("Age list:", age_list)
print("Name list:", name_list)

# 2. List Operations
print("\n--- Operations ---")
name_list.append("Yazhini")
print("a. After Append:", name_list)

age_list.insert(2, 30)
print("b. After Insert 30 at index 2:", age_list)

name_list.remove("Yazhini")
print("c. After Remove Yazhini:", name_list)

age_list.pop()
print("d. After Pop last:", age_list)

age_list.extend([29, 30, 26])
print("e. After Extend:", age_list)

age_list.sort(reverse=True)
print("f. After Sort Descending:", age_list)

print("g. Max age:", max(age_list))
print("g. Min age:", min(age_list))
print("g. Sum of ages:", sum(age_list))

# 3. Accessing List Elements
print("\n--- Accessing ---")
print("a. First element:", name_list[0])
print("b. Last element:", name_list[-1])
print("c. Index 2 to 4:", name_list[2:5])
print("d. Reverse:", name_list[::-1])
```

```python
# Dictionary
student_marks = {
    "Anjali": 95,
    "Meera": 88,
    "Asha": 76,
    "Divya": 90,
    "Lakshmi": 85
}

print("a. Student Marks:", student_marks)

# b. Access one student
print("b. Mark of Anjali:", student_marks["Anjali"])

# c. Add Janani
student_marks["Janani"] = 80
print("c. After Adding Janani:", student_marks)

# d. Update one old student to 82
student_marks["Asha"] = 82
print("d. After Updating Asha to 82:", student_marks)

# e. keys, values, items
print("e. Keys:", list(student_marks.keys()))
print("e. Values:", list(student_marks.values()))
print("e. Items:", list(student_marks.items()))
```

```python
# Sets

# a. Create set
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print("a. my_set:", my_set)
print("Explanation: Set-il duplicate values varilla. 'a' 3 thavana, 'i' 2 thavana koduthittum orikkal mathrame varullu. Athukond 5 unique values mathrame kittullu.")

# b. Try to change my_set[4] = 's'
print("\n--- b ---")
try:
    my_set[4] = 's'
except TypeError as e:
    print("Error:", e)
    print("Explanation: Set-il indexing nadakkilla. Set unordered aanu, athukond my_set[4] ennu paranjal eth index aanennu Python-nu ariyilla. Set mutable aanu, pakshe index vazhi maattanam pattilla. add() / remove() mathrame pattullu.")

# c. Two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("\n c. set1:", set1)
print("c. set2:", set2)

# d. Union and Intersection
print("d. Union:", set1.union(set2))
print("d. Intersection:", set1.intersection(set2))
```
